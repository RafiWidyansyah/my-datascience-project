# Import Library

```python
from google_play_scraper import app, reviews
import pandas as pd
```

# Scraping All Review From Game App "Roblox"

```python
scrapreview, _ = reviews(
    'com.roblox.client',
    lang='id',
    country='id',
    count=10000
)
```

```python
import csv

with open('ulasan_game_roblox.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Review'])
    for review in scrapreview:
        writer.writerow([review['content']])
```

# Load Dataset

```python
game_reviews_df = pd.DataFrame(scrapreview)
game_reviews_df.shape
game_reviews_df.head()
game_reviews_df.to_csv('ulasan_game_roblox.csv', index=False)
```
